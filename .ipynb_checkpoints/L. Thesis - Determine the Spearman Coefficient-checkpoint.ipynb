##### Function to create Kendall matrix

```python
def get_kendall_score(feature_importances):
    feature_importances_values = feature_importances.drop(columns=['feature'])

    # Convert importances to ranks (highest importance -> rank 1)
    feature_importances_ranks_df = feature_importances_values.rank(ascending=False)

    # Compute pairwise Kendall tau for all XAI methods
    methods = feature_importances_ranks_df.columns
    kendall_matrix = pd.DataFrame(index=methods, columns=methods, dtype=float)

    for i in methods:
        for j in methods:
            tau, _ = kendalltau(feature_importances_ranks_df[i], feature_importances_ranks_df[j])
            kendall_matrix.loc[i, j] = tau

    upper_triangle = kendall_matrix.where(np.triu(np.ones(kendall_matrix.shape), k=1).astype(bool))
    mean_kendall = upper_triangle.stack().mean()
    return kendall_matrix, mean_kendall
```
