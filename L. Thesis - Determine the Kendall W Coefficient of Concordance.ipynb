##### Function to create Kendall Matrix and Score

```python
def kendalls_w(feature_importances_df):
    """
    Compute Kendall’s W (Coefficient of Concordance).

    Parameters
    ----------
    ranks_df : pd.DataFrame
        Rows are items (e.g., features),
        columns are judges/methods,
        values are ranks (1 = highest importance).

    Returns
    -------
    float
        Kendall's W in [0, 1]
    """
    feature_importances_values = feature_importances_df.drop(columns=['feature'])
    
    # Convert importances to ranks (highest importance -> rank 1)
    ranks_df = feature_importances_values.rank(ascending=False)
    
    # Number of items (n) and judges (m)
    n, m = ranks_df.shape

    # Sum of ranks for each item
    R = ranks_df.sum(axis=1)

    # Mean of rank sums
    R_bar = R.mean()

    # Sum of squared deviations
    S = ((R - R_bar) ** 2).sum()

    # Kendall's W formula
    W = 12 * S / (m**2 * (n**3 - n))

    return W
```
