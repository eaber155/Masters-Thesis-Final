```python
lime_explainer = lime.lime_tabular.LimeTabularExplainer(
    training_data=X_train_scaled.values,
    feature_names=feature_names,
    mode='classification',
    class_names=["non-fraud", "fraud"],
    verbose=True
)
```

```python
i = 0 # Choose an index to explain
sample = X_test_scaled.values[i]
```

```python
# Wrap decision_function into a "pseudo-probability"
def predict_proba_iso(X):
    # Ensure X has feature names if iso_forest was trained with a DataFrame
    if isinstance(X, np.ndarray):
        X = pd.DataFrame(X, columns=X_train_scaled.columns)  # use training feature names

    scores = iso_forest.decision_function(X)  # higher = more normal

    # Normalize scores into [0, 1] for pseudo-probability
    probs = (scores - scores.min()) / (scores.max() - scores.min() + 1e-9)

    # Two-class probabilities: [anomaly, normal]
    return np.vstack([1 - probs, probs]).T
```

```python
lime_exp = lime_explainer.explain_instance(
    data_row=X_test_scaled.iloc[i],
    predict_fn=predict_proba_iso,
    num_features=10
)
```

#### 5.3 Applying Counterfactual to the Isolation Forest Model
* E2A. Thesis - Test different XAI methods with the Isolation Forest Model (Counterfactual)

```python
from sklearn.metrics import pairwise_distances
```

```python
# Show floats with 2 decimal places and avoid scientific notation
pd.set_option('display.float_format', '{:.4f}'.format)
```

```python
# Assume: X_train, X_test, model
query_instance = X_test.iloc[idx].copy()
query_instance_values = query_instance.values
```

```python
def predict_label(model, x):
    """
    Predict class for a single instance.
    """
    if hasattr(model, "predict_proba"):
        return model.predict_proba(x.reshape(1, -1))[0,1]  # probability of class 1
    else:
        return model.predict(x.reshape(1, -1))[0]
```

```python
def distance(x1, x2):
    """Euclidean distance between two instances"""
    return np.linalg.norm(x1 - x2)
```

```python
def generate_counterfactual(model, x0, total_cfs=3, max_trials=10000, step_size=0.1):
    """
    Generates counterfactuals using random perturbations.
    """
    cf_list = []
    trials = 0
   
    while len(cf_list) < total_cfs and trials < max_trials:
        # Perturb features randomly
        x_cf = x0 + np.random.normal(0, step_size, size=x0.shape)
       
        # Check if prediction flips
        pred_orig = predict_label(model, x0)
        pred_cf = predict_label(model, x_cf)
       
        # For binary classification, flip the label
        if (pred_orig < 0.5 and pred_cf >= 0.5) or (pred_orig >= 0.5 and pred_cf < 0.5):
            cf_list.append(x_cf)
       
        trials += 1
   
    return np.array(cf_list)
```

```python
start_time = time.time()
```

```python
counterfactuals = generate_counterfactual(model=iso_forest, x0=query_instance_values, total_cfs=3)
```

```python
end_time = time.time()
explanation_time = end_time - start_time
print(f"Counterfactual on Isolation Forest (Explanation Time): {explanation_time:.2f} seconds")
```

```python
counterfactuals.shape[0]
```

```python
if counterfactuals.shape[0] == 0:
    print("couldn’t find any valid alternative instance for the instance with index {idx}")
else:
    # Create DataFrame for counterfactual(s)
    df_cf = pd.DataFrame(counterfactuals, columns=X_test_scaled.columns)
    
    # Combine original and counterfactual without hierarchical index
    df_compare = pd.concat([query_instance.to_frame().T, df_cf], axis=0)
    df_compare.index = ['Original'] + [f'Counterfactual_{i+1}' for i in range(len(df_cf))]
    
    # Transpose so features are rows (no hierarchical index)
    df_transposed = df_compare.T
    
    # Compute delta with first counterfactual (or modify for multiple)
    df_transposed['delta'] = df_transposed['Counterfactual_1'] - df_transposed['Original']
    df_transposed['abs_delta'] = np.abs(df_transposed['delta'])
    
    # Sort features by absolute change and pick top 10
    df_top10 = df_transposed.sort_values('abs_delta', ascending=False).head(10)
    
    # Optionally, round for readability
    df_top10 = df_top10.round(4)
    
    print("Top 10 feature changes:")
    print(df_top10)
```

```python
if counterfactuals.shape[0] == 0:
    print("Couldn't find any valid alternative instance for the instance with index {idx}")
else:
    # Horizontal bar plot of absolute changes
    plt.figure(figsize=(10, 6))
    sns.barplot(
        x='abs_delta',
        y=df_top10.index,
        data=df_top10,
        palette='viridis', 
        hue=df_top10.index
    )
    plt.xlabel('Absolute Change')
    plt.ylabel('Feature')
    plt.title('Top 10 Feature Changes in Counterfactual - XGBoost')
    plt.tight_layout()
    plt.show()
```

```python
if counterfactuals.shape[0] == 0:
    print("Couldn't find any valid alternative instance for the instance with index {idx}")
else:
    # heatmap to see original vs counterfactual values
    plt.figure(figsize=(10, 6))
    sns.heatmap(
        df_top10[['Original', 'Counterfactual_1']],
        annot=True,
        fmt=".4f",
        cmap='coolwarm'
    )
    plt.title('Original vs Counterfactual Values for Top 10 Features - XGBoost')
    plt.ylabel('Feature')
    plt.show()
```
