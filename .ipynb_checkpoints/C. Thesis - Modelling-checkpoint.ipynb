### 2.2 Split the data into Train-Test

#### 2.2.1 Create a function to split the data so as to have 5% bad credits in the test data

```python
def split_with_realistic_test(X, y, test_size=0.2, bad_class=1, bad_fraction_test=0.05, random_state=1):
    """
    Splits X, y into train/test so that the test set has a realistic prevalence of the minority class.

    Parameters:
    - X: pd.DataFrame, features
    - y: pd.Series, target
    - test_size: float, fraction of total data to use as test set
    - bad_class: int, label of the minority class
    - bad_fraction_test: float, desired fraction of minority in test set
    - random_state: int, reproducibility

    Returns:
    - X_train, X_test, y_train, y_test
    """
    # Separate minority and majority
    X_bad = X[y == bad_class]
    y_bad = y[y == bad_class]

    X_good = X[y != bad_class]
    y_good = y[y != bad_class]

    n_test = int(len(X) * test_size)
    n_bad_test = int(n_test * bad_fraction_test)
    n_good_test = n_test - n_bad_test

    # Sample for test set
    X_bad_test, X_bad_rest, y_bad_test, y_bad_rest = train_test_split(
        X_bad, y_bad, test_size=len(X_bad) - n_bad_test, random_state=random_state
    )
    X_good_test, X_good_rest, y_good_test, y_good_rest = train_test_split(
        X_good, y_good, test_size=len(X_good) - n_good_test, random_state=random_state
    )

    # Combine test and train sets
    X_test = pd.concat([X_bad_test, X_good_test]).sample(frac=1, random_state=random_state).reset_index(drop=True)
    y_test = pd.concat([y_bad_test, y_good_test]).sample(frac=1, random_state=random_state).reset_index(drop=True)

    X_train = pd.concat([X_bad_rest, X_good_rest]).sample(frac=1, random_state=random_state).reset_index(drop=True)
    y_train = pd.concat([y_bad_rest, y_good_rest]).sample(frac=1, random_state=random_state).reset_index(drop=True)

    return X_train, X_test, y_train, y_test
```

#### 2.2.2 Separate the features and the label

```python
X = dataset.drop(columns=['credit_risk'])
y = dataset['credit_risk']
```

#### 2.2.3 Split the data using the custom function

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = split_with_realistic_test(X, y, test_size=0.2, bad_fraction_test=0.05)
```

```python
# Count the positive class
n_positive = (y_test == 1).sum()

# Total number of samples
n_total = len(y_test)

# Percentage of positive class
positive_percentage = (n_positive / n_total) * 100

print(f"Percentage of positive (bad credits) class in test set: {positive_percentage:.0f}%")
```

### 2.3 Feature Scaling

#### 2.3.1 Checking if the data has outliers

```python
# Write a function to detect outliers using IQR method for each column
def detect_iqr_outliers(df):
    outlier_flags = pd.DataFrame(False, index=df.index, columns=df.columns)

    # Use the inter-quartile range to calculate outliers
    for col in df.select_dtypes(include=[int, float]).columns:
        Q1 = df[col].quantile(0.25)
        Q3 = df[col].quantile(0.75)
        IQR = Q3 - Q1

        # Define bounds
        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR

        # Mark data points as false if not outliers and as true if outliers
        outlier_flags[col] = (df[col] < lower_bound) | (df[col] > upper_bound)

    return outlier_flags
```

```python
# Call the function
outliers_df = detect_iqr_outliers(dataset)
outliers_df.head()
```

```python
# For each column, count number of data points marked as false (not outliers) and as true (outliers)
outlier_count_df = outliers_df.apply(lambda col: col.value_counts()).T
```

```python
# Extract results in to a CSV file
outlier_count_df.to_csv('outlier_count_df.csv')
```

#### 3.4.2 Autoencoder

```python
import tensorflow as tf
from tensorflow.keras import layers, models
```

```python
# NOTE: Training is done on only on non-fraudulent samples
X_train_ae = X_train_scaled[y_train == 0] 
```

```python
# Build the autoencoder
input_dim = X_train_ae.shape[1]
encoding_dim = 32  # compressed representation size

input_layer = layers.Input(shape=(input_dim,))
encoded = layers.Dense(encoding_dim, activation='relu')(input_layer)
encoded = layers.Dense(16, activation='relu')(encoded)

decoded = layers.Dense(encoding_dim, activation='relu')(encoded)
decoded = layers.Dense(input_dim, activation='linear')(decoded)

autoencoder = models.Model(inputs=input_layer, outputs=decoded)

autoencoder.compile(optimizer='adam', loss='mse')
```

```python
start_time = time.time()
```

```python
# Train the autoencoder
history = autoencoder.fit(
    X_train_ae, X_train_ae,
    epochs=50,
    batch_size=256,
    shuffle=True,
    validation_split=0.1,
    verbose=0
)
```

```python
end_time = time.time()
training_time = end_time - start_time
print(f"Autoencoder Training Time: {training_time:.2f} seconds")
```

```python
# Compute reconstruction errors on test data
X_test_pred = autoencoder.predict(X_test_scaled)
mse = np.mean(np.power(X_test_scaled - X_test_pred, 2), axis=1)
```

```python
# Evaluate anomaly detection performance
# A simple threshold for anomaly score
threshold = np.percentile(mse[y_test == 0], 95)  # 95th percentile of reconstruction error on non-fraud
```

```python
# Predict fraud if reconstruction error > threshold
y_pred = (mse > threshold).astype(int)
```

```python
y_pred_proba = mse
```

```python
fpr, tpr, thresholds = roc_curve(y_test, y_pred_proba,pos_label=1)
auc_score = auc(fpr, tpr)
print("Autoencoder AUC (in %):", auc_score*100)
```

```python
# ROC-AUC Curve
# plot roc curves
random_probs = [0 for i in range(len(y_test))]
p_fpr, p_tpr, thresholds = roc_curve(y_test, random_probs, pos_label=1)
plt.plot(fpr, tpr, linestyle='--', color='orange')
plt.plot(p_fpr, p_tpr, linestyle='--', color='blue')
# title
plt.title('Autoencoder ROC curve')
# x label
plt.xlabel('False Positive Rate')
# y label
plt.ylabel('True Positive rate')

plt.savefig('Autoencoder ROC curve',dpi=300)
plt.show()
```

```python
### Calculate recall_score of the Autoencoder to determine its sensitivity

sensitivity = recall_score(y_test, y_pred, pos_label=1)
print(f"Autoencoder Recall Score (Sensitivity): {sensitivity:.4f}")
```

```python
### Calculate precision_score of the Autoencoder

precision = precision_score(y_test, y_pred, pos_label=1)
print(f"Autoencoder Precision Score: {precision:.4f}")
```

#### 3.5.3 Artificial Neural Network (ANN)

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping
```

```python
# Build an ANN model
model = Sequential([
    Input(shape=(X_train.shape[1],)),  # Explicit Input layer instead of input_dim in Dense
    Dense(64, activation='relu'),
    Dropout(0.3),
    Dense(32, activation='relu'),
    Dropout(0.2),
    Dense(1, activation='sigmoid')
])

model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['AUC'])

# Early stopping to prevent overfitting
early_stop = EarlyStopping(monitor='val_AUC', patience=3, restore_best_weights=True, mode='max')
```

```python
start_time = time.time()
```

```python
# Train the ANN model with timing
history = model.fit(
    X_train_scaled, y_train,
    validation_split=0.2,
    epochs=20,
    batch_size=256,
    callbacks=[early_stop],
    verbose=0
)
```

```python
end_time = time.time()
training_time = end_time - start_time
print(f"ANN Training Time: {training_time:.2f} seconds")
```

```python
# Predict y given X_test_scaled
y_pred_proba = model.predict(X_test_scaled)
y_pred = (y_pred_proba > 0.5).astype(int)
```

```python
fpr, tpr, thresholds = roc_curve(y_test, y_pred_proba,pos_label=1)
auc_score = auc(fpr, tpr)
print("ANN AUC (in %):", auc_score*100)
```
