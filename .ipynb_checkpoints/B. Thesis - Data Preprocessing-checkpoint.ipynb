### Step 1: Preliminary check for missing data

```python
def top_missing_cols_perc(df):
    """
    returns missing columns in dataframe with missing values percent > thresh
    if n=None. It will gave whole dataframe with missing values percent > thresh
    """
    
    dff = (df.isnull().sum()/df.shape[0])*100
    dff = dff.reset_index()
    dff.columns = ['col','missing_percent']
    dff = dff.sort_values(by=['missing_percent'],ascending=True).reset_index(drop=True)
    return dff
```

```python
missing_perc_matrix = top_missing_cols_perc(dataset)
```

```python
print("Number of columns with missing data: "+ str(missing_perc_matrix[missing_perc_matrix['missing_percent']>0.00].shape[0]) + " columns")
```

### Step 2: Encoding Categorical Variables

```python
all_features = dataset.columns.to_list()

columns = [
   'status', 'duration', 'credit_history', 'purpose', 'amount', 'savings', 'employment_duration', 'installment_rate', 'personal_status_sex',
    'other_debtors', 'present_residence', 'property', 'age', 'other_installment_plans', 'housing', 'number_credits', 'job', 'people_liable',
    'telephone', 'foreign_worker', 'credit_risk'
]

# locate and load the file
dataset_encoded = pd.read_csv("SouthGermanCredit.asc", sep=r"\s+", header=0, names=columns)

# Remap the label field such that bad credits in the positive class
dataset_encoded["credit_risk"] = dataset_encoded["credit_risk"].map({0: 1, 1:0})
#print(df.head())
#print(df.info())
dataset_encoded.head()
```

### Step 3: Addressing multicollinearity among features

#### 3.1 Correlation Analysis

```python
corr_matrix = dataset_encoded.corr()
```

```python
mask = np.triu(np.ones(corr_matrix.shape), k=1).astype(bool)
item_pairs = corr_matrix.where(mask)
```

```python
correlation_df = item_pairs.stack().reset_index()
correlation_df.columns = ['Feature 1', 'Feature 2', 'Correlation']
```

```python
correlation_df.head()
```

```python
#correlation_df['Correlation'] = correlation_df['Correlation'].apply(lambda x: x*100)
correlation_df['Abs_Correlation'] = correlation_df['Correlation'].apply(lambda x: abs(x))
```

##### 3.1.1 Separate label-related (credit_risk) correlation

```python
def feature_(cols):
    feature1 = cols['Feature 1']
    feature2 = cols['Feature 2']
    if feature1=='credit_risk':
        return feature2
    else:
        return feature1
```

```python
label_corr = correlation_df[(correlation_df['Feature 1']=='credit_risk')|(correlation_df['Feature 2']=='credit_risk')].copy()
label_corr['Feature'] = label_corr.apply(feature_,axis=1)
```

```python
label_corr = label_corr.drop(columns=['Feature 1', 'Feature 2'])
```

```python
label_corr.rename(columns={'Correlation':'label_corr', 'Abs_Correlation':'label_abs_corr'},inplace=True)
label_corr
```

##### 3.1.2 Get a Feature only table and add correlation to credit_risk on the feature tables

```python
features_corr = correlation_df[(correlation_df['Feature 1']!='credit_risk')&(correlation_df['Feature 2']!='credit_risk')]
features_corr = features_corr.merge(label_corr[['Feature','label_abs_corr']],left_on=['Feature 1'], right_on=['Feature'], how='left')
```

```python
features_corr.rename(columns={'label_abs_corr':'label_abs_corr_feat1'},inplace=True)
features_corr = features_corr.drop(columns=['Feature'])
```

```python
features_corr = features_corr.merge(label_corr[['Feature','label_abs_corr']],left_on=['Feature 2'], right_on=['Feature'], how='left')
```

```python
features_corr.rename(columns={'label_abs_corr':'label_abs_corr_feat2'},inplace=True)
features_corr = features_corr.drop(columns=['Feature'])
```

##### 3.1.3 Pick Combinations with correlations higher than 0.7

```python
features_corr_v70 = features_corr[features_corr['Abs_Correlation']>0.7].copy()
```

```python
def features_ABOVE70_check(df):
    if df.shape[0]==0:
        print("No feature pairs with correlation above 0.7 (i.e. 70%)")
    else:
        print(str(df.shape[0]) + " features with above 0.7 (i.e. 70%)")
```

```python
features_ABOVE70_check(features_corr_v70)
```
